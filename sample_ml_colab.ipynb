```python
for name in os.listdir("."):
    for name in os.listdir("."):
        details = os.stat(name)
        kind = "Directory" if os.path.isdir(name) else "File"
        print(f"{name} | {kind} | {details.st_size} bytes | Modified: {details.st_mtime}")
```

```python
for entry in os.scandir("sample_data"):
    details = entry.stat()
    kind = "Directory" if entry.is_dir() else "File"
    print(f"{entry.name} | {kind} | {details.st_size} bytes")
```
