# Download the Reddit Sarcasm dataset in Colab

Run these cells in Google Colab to download the dataset into the Colab runtime at `/content/reddit-sarcasm`. The dataset is about 3 GB, so the first download may take a while. It is not downloaded to your computer. Colab runtime files are temporary and are removed when the runtime is reset.

```python
%pip -q install huggingface_hub
```

```python
from pathlib import Path
from huggingface_hub import hf_hub_download

dataset_file = hf_hub_download(
    repo_id="Thewillonline/reddit-sarcasm",
    filename="reddit-sarcasm.txt",
    repo_type="dataset",
    local_dir="/content/reddit-sarcasm",
)

file_size_gb = Path(dataset_file).stat().st_size / 1_000_000_000
print(f"Downloaded to: {dataset_file}")
print(f"File size: {file_size_gb:.2f} GB")
```
